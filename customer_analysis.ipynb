```python
# 1. Drop the duplicate column
df.drop(columns=['Promo Code Used'], inplace=True)

# 2. Standardise Frequency labels
df['Frequency of Purchases'] = df['Frequency of Purchases'].replace({
    'Every 3 Months': 'Quarterly',
    'Bi-Weekly': 'Fortnightly'
})

# 3. Fill missing Review Ratings with the median
median_rating = df['Review Rating'].median()
df['Review Rating'].fillna(median_rating, inplace=True)

# Confirm
print("Columns now:", df.shape[1])
print("Nulls remaining:", df.isnull().sum().sum())
print("Frequency values now:", df['Frequency of Purchases'].unique())
print("Median rating used for fill:", median_rating)
```

```python
frequency_map = {
    'Weekly': 5,
    'Fortnightly': 4,
    'Monthly': 3,
    'Quarterly': 2,
    'Annually': 1
}

df['Frequency Score'] = df['Frequency of Purchases'].map(frequency_map)

print(df[['Frequency of Purchases', 'Frequency Score']].drop_duplicates().sort_values('Frequency Score', ascending=False))
```

```python
# Normalize both columns to 0-1 scale
df['Amount Normalized'] = (df['Purchase Amount (USD)'] - df['Purchase Amount (USD)'].min()) / \
                           (df['Purchase Amount (USD)'].max() - df['Purchase Amount (USD)'].min())

df['Purchases Normalized'] = (df['Previous Purchases'] - df['Previous Purchases'].min()) / \
                               (df['Previous Purchases'].max() - df['Previous Purchases'].min())

# Value Score = equal weight on spend and purchase history
df['Value Score'] = (df['Amount Normalized'] + df['Purchases Normalized']) / 2

# Cut into 3 tiers
df['Value Tier'] = pd.cut(df['Value Score'],
                           bins=3,
                           labels=['Low Value', 'Mid Value', 'High Value'])

print(df['Value Tier'].value_counts())
print("\nAverage Value Score per tier:")
print(df.groupby('Value Tier')['Value Score'].mean().round(3))
```

```python
# --- Definition A: Behaviour-based ---
df['Purchases Freq Normalized'] = (df['Frequency Score'] - 1) / (5 - 1)

df['Loyalty Score A'] = (df['Purchases Normalized'] + df['Purchases Freq Normalized']) / 2

# --- Definition B: Discount-adjusted ---
# Penalty: if customer always uses discount, reduce loyalty score by 20%
discount_penalty = df['Discount Applied'].apply(lambda x: 0.8 if x == 'Yes' else 1.0)

df['Loyalty Score B'] = df['Loyalty Score A'] * discount_penalty

# Compare both
print("Loyalty Score A (behaviour only):")
print(df['Loyalty Score A'].describe().round(3))
print("\nLoyalty Score B (discount-adjusted):")
print(df['Loyalty Score B'].describe().round(3))
print("\nCorrelation between A and B:", df['Loyalty Score A'].corr(df['Loyalty Score B']).round(3))
```

```python
# Subscribers always get discounts - they are structurally dependent
# Non-subscribers who still used discount = opportunistic dependent
# No discount at all = independent

def dependency_score(row):
    if row['Subscription Status'] == 'Yes':
        return 2   # structural dependency - locked into discount model
    elif row['Discount Applied'] == 'Yes':
        return 1   # opportunistic - bought because of a deal
    else:
        return 0   # independent - bought without any discount

df['Promo Dependency Score'] = df.apply(dependency_score, axis=1)

print(df['Promo Dependency Score'].value_counts())
print("\nLabel breakdown:")
print("0 = Independent, 1 = Opportunistic, 2 = Structurally Dependent")
```

```python
satisfaction_threshold = df['Review Rating'].median()

df['Satisfaction Flag'] = df['Review Rating'].apply(
    lambda x: 'Satisfied' if x > satisfaction_threshold else 'Unsatisfied'
)

print("Threshold used:", satisfaction_threshold)
print(df['Satisfaction Flag'].value_counts())
print("\nAvg Purchase Amount by Satisfaction:")
print(df.groupby('Satisfaction Flag')['Purchase Amount (USD)'].mean().round(2))
print("\nAvg Previous Purchases by Satisfaction:")
print(df.groupby('Satisfaction Flag')['Previous Purchases'].mean().round(2))
```

```python
df['Loyalty Tier'] = pd.cut(df['Loyalty Score B'],
                              bins=3,
                              labels=['Low Loyalty', 'Mid Loyalty', 'High Loyalty'])

print(df['Loyalty Tier'].value_counts())

print("\nMean Loyalty Score B per tier:")
print(df.groupby('Loyalty Tier')['Loyalty Score B'].mean().round(3))

print("\nMean Previous Purchases per tier:")
print(df.groupby('Loyalty Tier')['Previous Purchases'].mean().round(1))

print("\nMean Frequency Score per tier:")
print(df.groupby('Loyalty Tier')['Frequency Score'].mean().round(2))
```

```python
df.to_csv('customer_analysis_clean.csv', index=False)

print("Saved successfully.")
print("Final columns:", df.columns.tolist())
print("Final shape:", df.shape)
```
